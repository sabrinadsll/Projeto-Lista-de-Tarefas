### **Desafio 2: Ajustar nomes de colunas**

Após carregar os dados, ajuste os nomes das colunas para garantir que não haja espaços ou caracteres especiais. Por exemplo, a coluna "UF Autor" deve ser transformada em "uf_autor".

```python
df.rename(columns={
    'Ano':'ano',
    'RP':'rp',
    'Autor':'autor',
    'Tipo Autor':'tipo_autor',
    'Partido':'partido',
    'UF Autor':'uf_autor',
    'Nro. Emenda':'num_emenda',
    'Órgão':'orgao',
    'Ação':'acao',
    'Dotação Inicial Emenda':'dotacao_inicial_emenda',
    'Dotação Atual Emenda':'dotacao_atual_emenda',
    'Empenhado':'empenhado',
    'Liquidado':'liquidado',
    'Pago':'pago'
}, inplace=True)
```

### **Desafio 3: Ajustar valores em algumas colunas**

Melhore a apresentação de dados em algumas colunas, por exemplo, separe o número do texto da coluna "Órgão". Descreva as operações que você fez ao finalizar esta etapa.

**Importante:** As colunas 'Dotação Inicial', 'Emenda','Dotação Atual Emenda',	'Empenhado',	'Liquidado' e	'Pago' precisam ser tratadas. Retire os "." e transforme em *float* estas colunas.

```python
df['orgao'].head()
```

```python
df['codigo_orgao'] = df['orgao'].apply(
    lambda x: x.split(' - ')[0]
)
```

criei a coluna codigo_orgao, depois peguei cada valor da coluna orgao e
separei o texto usando o split onde tinha o -, depois usei o índice 0 para pegar a primeira parte

```python
df['nome_orgao'] = df['orgao'].apply(
    lambda x: x.split(' - ')[1]
)
```

```python
df[['orgao', 'codigo_orgao', 'nome_orgao']].head()
```

```python
df['dotacao_inicial_emenda'] = df['dotacao_inicial_emenda'].apply(
    lambda x: x.replace('.', '')
)

df['dotacao_atual_emenda'] = df['dotacao_atual_emenda'].apply(
    lambda x: x.replace('.', '')
)

df['empenhado'] = df['empenhado'].apply(
    lambda x: x.replace('.', '')
)

df['liquidado'] = df['liquidado'].apply(
    lambda x: x.replace('.', '')
)

df['pago'] = df['pago'].apply(
    lambda x: x.replace('.', '')
)
```

```python
df = df.astype({
    'dotacao_inicial_emenda':'float',
    'dotacao_atual_emenda':'float',
    'empenhado':'float',
    'liquidado':'float',
    'pago':'float'
})
```
